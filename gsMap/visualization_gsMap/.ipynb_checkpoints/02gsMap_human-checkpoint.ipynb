```python
import os
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# =========================
# 参数
# =========================
BASE_DIR = "/biostack/home/suchenyi/gsmap"
REGIONS = ["mPFC", "DRN", "Stri", "HIP", "VTA"]
CONDITIONS = ["Control", "3D", "7D", "Res", "Sus"]
TRAITS = ["MDD"]

# ① 手动覆盖：自动对齐失败的样本，先额外旋转（逆时针为正）
MANUAL_OVERRIDES = {
    ("Stri", "7D"):  180,
    ("Stri", "Sus"): 180,
    ("mPFC","Res"):180
}

# ② 每个脑区的最终摆正角度（逆时针为正，顺时针用负数）
REGION_ROTATIONS = {
    "mPFC":  -90,
    "DRN":   -60,
    "Stri":  90,
    "HIP":   60,
    "VTA":   210,
}

# =========================
# 工具函数
# =========================
def load_ldsc(region, condition, trait):
    dfs = []
    for rep in ["rep1", "rep2"]:
        path = (f"{BASE_DIR}/{region}_{condition}_{rep}/spatial_ldsc/"
                f"{region}_{condition}_{rep}_{trait}.csv.gz")
        if not os.path.exists(path):
            continue
        df = pd.read_csv(path)
        df["spot"] = df["spot"].astype(str)
        dfs.append(df[["spot", "z"]].rename(columns={"z": f"z_{rep}"}))

    if not dfs:
        return None

    df_merge = dfs[0]
    for df in dfs[1:]:
        df_merge = df_merge.merge(df, on="spot", how="outer")

    z_cols = [c for c in df_merge.columns if c.startswith("z_")]
    df_merge["z_mean"] = df_merge[z_cols].mean(axis=1)
    return df_merge


def rotate_coords(coords: np.ndarray, angle_deg: float) -> np.ndarray:
    """绕质心逆时针旋转 angle_deg 度（顺时针传负值）。"""
    theta = np.radians(angle_deg)
    c, s = np.cos(theta), np.sin(theta)
    R = np.array([[c, -s], [s, c]])
    centroid = coords.mean(axis=0)
    return (coords - centroid) @ R.T + centroid


def align_to_reference(coords: np.ndarray,
                        ref_coords: np.ndarray,
                        grid_size: int = 60) -> np.ndarray:
    """枚举 8 种刚性变换，用网格重叠度选最优方向。"""

    def normalize(c):
        c = c.astype(float).copy()
        for d in range(2):
            lo, hi = c[:, d].min(), c[:, d].max()
            c[:, d] = (c[:, d] - lo) / max(hi - lo, 1e-9)
        return c

    def to_grid(c):
        g = np.zeros((grid_size, grid_size), dtype=np.float32)
        xi = np.clip((c[:, 0] * (grid_size - 1)).astype(int), 0, grid_size - 1)
        yi = np.clip((c[:, 1] * (grid_size - 1)).astype(int), 0, grid_size - 1)
        g[yi, xi] = 1.0
        return g

    def get_8_transforms(c):
        c = c.astype(float).copy()
        c[:, 0] -= c[:, 0].min()
        c[:, 1] -= c[:, 1].min()
        x, y = c[:, 0], c[:, 1]
        xm, ym = x.max(), y.max()
        return [
            np.c_[x,       y      ],
            np.c_[xm - x,  y      ],
            np.c_[x,       ym - y ],
            np.c_[xm - x,  ym - y ],
            np.c_[y,       x      ],
            np.c_[ym - y,  x      ],
            np.c_[y,       xm - x ],
            np.c_[ym - y,  xm - x ],
        ]

    ref_grid = to_grid(normalize(ref_coords))
    best_coords, best_score = coords, -np.inf
    for candidate in get_8_transforms(coords):
        score = float(np.sum(ref_grid * to_grid(normalize(candidate))))
        if score > best_score:
            best_score = score
            best_coords = candidate
    return best_coords


# =========================
# 画图
# =========================
reference_coords = {}

fig, axes = plt.subplots(
    nrows=len(REGIONS),
    ncols=len(CONDITIONS),
    figsize=(15, 15),
    constrained_layout=True,
)

for i, region in enumerate(REGIONS):
    for j, condition in enumerate(CONDITIONS):
        ax = axes[i, j]
        print(f"Processing {region} - {condition}")

        # ---- 读取 h5ad
        h5ad_path = (f"{BASE_DIR}/{region}_{condition}_rep1/"
                     f"{region}_{condition}_rep1_annotated.h5ad")
        if not os.path.exists(h5ad_path):
            ax.axis("off")
            continue

        adata = sc.read_h5ad(h5ad_path)
        adata.obs.index = adata.obs.index.astype(str)

        # ---- 读取 trait z-score
        trait_series = []
        for trait in TRAITS:
            df_ldsc = load_ldsc(region, condition, trait)
            if df_ldsc is None:
                continue
            trait_series.append(df_ldsc.set_index("spot")["z_mean"])

        if not trait_series:
            ax.axis("off")
            continue

        z_all = pd.concat(trait_series, axis=1)
        z_all["z"] = z_all.mean(axis=1)

        # ---- spot 对齐
        adata.obs["spot_clean"] = adata.obs.index.str.replace("-1", "", regex=False)
        adata.obs = adata.obs.merge(
            z_all[["z"]], left_on="spot_clean", right_index=True, how="left"
        )
        adata = adata[~adata.obs["z"].isna()].copy()

        # ---- 坐标流水线：平移 → 脑区内对齐 → 手动覆盖 → 脑区摆正
        coords = adata.obs[["x", "y"]].values.astype(float)
        coords[:, 0] -= coords[:, 0].min()
        coords[:, 1] -= coords[:, 1].min()

        # Step 1：脑区内对齐（以 Control 为参考）
        if condition == "Control":
            reference_coords[region] = coords.copy()
        else:
            ref = reference_coords.get(region)
            if ref is not None:
                coords = align_to_reference(coords, ref)

        # Step 2：手动补偿个别样本的对齐偏差
        override_angle = MANUAL_OVERRIDES.get((region, condition))
        if override_angle is not None:
            coords = rotate_coords(coords, override_angle)

        # Step 3：整个脑区统一摆正
        final_angle = REGION_ROTATIONS.get(region, 0)
        if final_angle != 0:
            coords = rotate_coords(coords, final_angle)

        adata.obsm["X_spatial_temp"] = coords

        # ---- 绘制子图
        sc.pl.embedding(
            adata,
            basis="X_spatial_temp",
            color="z",
            color_map="RdBu_r",
            vmin=-8,
            vmax=8,
            size=10,
            ax=ax,
            show=False,
            colorbar_loc=None,
             title="",
        )
        for text in ax.texts:
            if text.get_text() == "z":
                text.remove()

        # ③ 去掉边框、刻度、坐标轴标签
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_xlabel("")
        ax.set_ylabel("")
        for spine in ax.spines.values():
            spine.set_visible(False)

        # 保留行列标题（脑区名 / 条件名）
        if i == 0:
            ax.set_title(condition, fontsize=12, pad=4)
        if j == 0:
            ax.set_ylabel(region, fontsize=12, labelpad=4)

        ax.invert_yaxis()

# ---- 全局 colorbar
norm = mcolors.Normalize(vmin=-3, vmax=3)
sm = cm.ScalarMappable(cmap="RdBu_r", norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, ax=axes, fraction=0.02, shrink=0.6, pad=0.02)
cbar.set_label("z-score (spatial LDSC)", fontsize=12)

plt.savefig("gsmap_5x5_spatial.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import os
import scanpy as sc
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist
from adjustText import adjust_text
from scipy.cluster.hierarchy import fcluster

# =========================
# 参数
# =========================
BASE_DIR = "/biostack/home/suchenyi/gsmap"

REGIONS = ["mPFC", "DRN", "Stri", "HIP", "VTA"]
CONDITIONS = ["Control", "3D", "7D", "Res", "Sus"]
all_region_results = {}
all_mdd_correlations = {}

def load_traits_from_file(file_path):
    with open(file_path, "r") as f:
        traits = [line.strip() for line in f if line.strip() != ""]
    
    print(f"Loaded {len(traits)} traits from file")
    return traits

TRAIT_FILE = "traits_list.txt"
ALL_TRAITS = load_traits_from_file(TRAIT_FILE)

# =========================
# 读取并合并rep
# =========================
def load_ldsc(region, condition, trait):
    dfs = []
    for rep in ["rep1", "rep2"]:
        file_path = f"{BASE_DIR}/{region}_{condition}_{rep}/spatial_ldsc/{region}_{condition}_{rep}_{trait}.csv.gz"
        if not os.path.exists(file_path):
            continue
        df = pd.read_csv(file_path)
        df["spot"] = df["spot"].astype(str)
        df = df[["spot", "z"]].rename(columns={"z": f"z_{rep}"})
        dfs.append(df)
    if len(dfs) == 0:
        return None
    df_merge = dfs[0]
    for df in dfs[1:]:
        df_merge = df_merge.merge(df, on="spot", how="outer")
    z_cols = [c for c in df_merge.columns if "z_" in c]
    df_merge["z_mean"] = df_merge[z_cols].mean(axis=1)
    return df_merge


# =========================
# 主流程
# =========================
for region in REGIONS:
    for condition in CONDITIONS:

        print(f"\nProcessing: {region} - {condition}")
        TRAITS = ALL_TRAITS

        h5ad_path = f"{BASE_DIR}/{region}_{condition}_rep1/{region}_{condition}_rep1_annotated.h5ad"
        if not os.path.exists(h5ad_path):
            continue

        adata = sc.read_h5ad(h5ad_path)
        adata.obs.index = adata.obs.index.astype(str)
        adata.obs["spot_clean"] = adata.obs.index.str.replace("-1", "", regex=False)

        trait_maps = {}
        common_spots = None

        for trait in TRAITS:
            df_ldsc = load_ldsc(region, condition, trait)
            if df_ldsc is None:
                continue
            df_ldsc = df_ldsc.set_index("spot")
            tmp = adata.obs[["spot_clean"]].merge(
                df_ldsc[["z_mean"]],
                left_on="spot_clean",
                right_index=True,
                how="inner"
            )
            if tmp.empty:
                continue
            z = tmp["z_mean"].values
            z = (z - np.mean(z)) / (np.std(z) + 1e-8)
            trait_maps[trait] = z
            if common_spots is None:
                common_spots = tmp.index
            else:
                common_spots = common_spots.intersection(tmp.index)

        if len(trait_maps) < 2:
            print("Not enough traits")
            continue

        traits = list(trait_maps.keys())
        X = np.vstack([trait_maps[t] for t in traits])
        n = len(traits)

        # =========================
        # 相似性矩阵（修复：动态尺寸 + 去掉annot + 缩小label字号）
        # =========================
        cell_size = 0.45                         # 每个格子占多少英寸
        fig_size = max(6, n * cell_size)         # 随性状数自动放大
        label_fontsize = max(4, min(9, 120 // n)) # 性状越多字越小

        corr_matrix = np.corrcoef(X)

        fig, ax = plt.subplots(figsize=(fig_size + 1, fig_size))
        sns.heatmap(
            corr_matrix,
            xticklabels=traits,
            yticklabels=traits,
            cmap="RdBu_r",
            center=0,
            annot=False,          # ← 性状多时关掉数字
            linewidths=0.3,
            linecolor="white",
            ax=ax,
            cbar_kws={"shrink": 0.6}
        )
        ax.set_xticklabels(ax.get_xticklabels(), rotation=90, fontsize=label_fontsize)
        ax.set_yticklabels(ax.get_yticklabels(), rotation=0,  fontsize=label_fontsize)
        ax.set_title(f"{region} - {condition}: Trait similarity", fontsize=11, pad=8)
        plt.tight_layout()
        plt.savefig(f"{region}_{condition}_trait_corr.png", dpi=300)
        plt.close()

        # =========================
        # 层次聚类（修复：动态图宽 + 旋转45° + 自适应字号）
        # =========================
        dist = pdist(X, metric="correlation")
        Z = linkage(dist, method="average")

        fig_w = max(8, n * 0.5)               # 每个叶节点至少 0.5 英寸
        dend_fontsize = max(5, min(10, 150 // n))

        fig, ax = plt.subplots(figsize=(fig_w, 5))
        dendrogram(
            Z,
            labels=traits,
            leaf_rotation=45,             # ← 旋转45°避免叠字
            leaf_font_size=dend_fontsize, # ← 自适应字号
            ax=ax
        )
        ax.set_title(f"{region} - {condition}: Trait clustering", fontsize=11)
        ax.set_ylabel("Distance")
        plt.tight_layout()
        plt.savefig(f"{region}_{condition}_trait_dendrogram.png", dpi=300)
        plt.close()

        # =========================
        # PCA（修复：adjustText 自动避让label重叠）
        # =========================
        pca = PCA(n_components=2)
        X_pca = pca.fit_transform(X)

        fig, ax = plt.subplots(figsize=(7, 6))
        ax.scatter(X_pca[:, 0], X_pca[:, 1], s=40, alpha=0.8)

        # 先把所有 text 对象收集起来，再统一调整
        texts = []
        for i, trait in enumerate(traits):
            texts.append(ax.text(X_pca[i, 0], X_pca[i, 1], trait, fontsize=7))

        adjust_text(
            texts,
            ax=ax,
            arrowprops=dict(arrowstyle="-", color="gray", lw=0.5),
            expand_points=(1.5, 1.5),   # 离点的最小扩展距离
            expand_text=(1.3, 1.3),     # 文字之间的最小扩展距离
        )

        ax.set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)")
        ax.set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)")
        ax.set_title(f"{region} - {condition}: Trait PCA", fontsize=11)
        ax.axhline(0, linestyle="--", linewidth=0.5, color="gray")
        ax.axvline(0, linestyle="--", linewidth=0.5, color="gray")

        plt.tight_layout()
        #plt.savefig(f"{region}_{condition}_trait_pca.png", dpi=300)
        plt.close()
        # =========================
        # 自动化性状提取与分类锁定 (新增部分)
        # =========================
# 1. 切割聚类
        from scipy.cluster.hierarchy import fcluster
        cluster_labels = fcluster(Z, t=2, criterion='maxclust')
        
        # 2. 构造当前脑区-状态的结果表
        current_df = pd.DataFrame({
            'Cluster': cluster_labels
        }, index=traits)
        
        if 'MDD' in traits:
            mdd_cluster_id = current_df.loc['MDD', 'Cluster']
            current_df[f'Is_MDD_Partner'] = (current_df['Cluster'] == mdd_cluster_id)
            
            # 记录相关系数，用于后续动态追踪
            mdd_idx = traits.index('MDD')
            all_mdd_correlations[f"{region}_{condition}"] = pd.Series(corr_matrix[mdd_idx], index=traits)
        
        all_region_results[f"{region}_{condition}"] = current_df

# 在主循环结束后运行
with pd.ExcelWriter("Brain_Region_Trait_Clustering_Results.xlsx") as writer:
    for key, df in all_region_results.items():
        # key 的格式是 "mPFC_Control" 等
        df.to_excel(writer, sheet_name=key[:31]) # Sheet名限长31字符
print("Excel 文件已保存：Brain_Region_Trait_Clustering_Results.xlsx")
```

```python
import os
import scanpy as sc
import pandas as pd
import numpy as np
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import pdist

BASE_DIR = "/biostack/home/suchenyi/gsmap"
REGIONS = ["mPFC", "DRN", "Stri", "HIP", "VTA"]
CONDITIONS = ["Control", "3D", "7D", "Res", "Sus"]

all_mdd_correlations = {}

def load_traits_from_file(file_path):
    with open(file_path, "r") as f:
        return [line.strip() for line in f if line.strip()]

ALL_TRAITS = load_traits_from_file("traits_list.txt")

def load_ldsc(region, condition, trait):
    dfs = []
    for rep in ["rep1", "rep2"]:
        file_path = f"{BASE_DIR}/{region}_{condition}_{rep}/spatial_ldsc/{region}_{condition}_{rep}_{trait}.csv.gz"
        if not os.path.exists(file_path):
            continue
        df = pd.read_csv(file_path)
        df["spot"] = df["spot"].astype(str)
        dfs.append(df[["spot", "z"]].rename(columns={"z": f"z_{rep}"}))
    if not dfs:
        return None
    df_merge = dfs[0]
    for df in dfs[1:]:
        df_merge = df_merge.merge(df, on="spot", how="outer")
    z_cols = [c for c in df_merge.columns if "z_" in c]
    df_merge["z_mean"] = df_merge[z_cols].mean(axis=1)
    return df_merge

# ── 主循环：只算相关矩阵，不画任何图 ─────────────────────────────
for region in REGIONS:
    for condition in CONDITIONS:
        print(f"Processing: {region} - {condition}", end=" ... ")

        h5ad_path = f"{BASE_DIR}/{region}_{condition}_rep1/{region}_{condition}_rep1_annotated.h5ad"
        if not os.path.exists(h5ad_path):
            print("skipped (no h5ad)")
            continue

        adata = sc.read_h5ad(h5ad_path)
        adata.obs.index = adata.obs.index.astype(str)
        adata.obs["spot_clean"] = adata.obs.index.str.replace("-1", "", regex=False)

        trait_maps = {}
        for trait in ALL_TRAITS:
            df_ldsc = load_ldsc(region, condition, trait)
            if df_ldsc is None:
                continue
            df_ldsc = df_ldsc.set_index("spot")
            tmp = adata.obs[["spot_clean"]].merge(
                df_ldsc[["z_mean"]], left_on="spot_clean", right_index=True, how="inner"
            )
            if tmp.empty:
                continue
            z = tmp["z_mean"].values
            z = (z - np.mean(z)) / (np.std(z) + 1e-8)
            trait_maps[trait] = z

        if len(trait_maps) < 2 or 'MDD' not in trait_maps:
            print("skipped (too few traits or no MDD)")
            continue

        traits = list(trait_maps.keys())
        X = np.vstack([trait_maps[t] for t in traits])

        # 只算相关矩阵，提取 MDD 那一行
        corr_matrix = np.corrcoef(X)
        mdd_idx = traits.index('MDD')
        all_mdd_correlations[f"{region}_{condition}"] = pd.Series(
            corr_matrix[mdd_idx], index=traits
        )
        print(f"done ({len(traits)} traits)")

print(f"\n完成！共 {len(all_mdd_correlations)} 个 Region-Condition")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

REGIONS    = ["mPFC", "DRN", "Stri", "HIP", "VTA"]
CONDITIONS = ["Control", "3D", "7D", "Res", "Sus"]

# ── 1. 宽表 & Top10 ───────────────────────────────────────────────
corr_df = pd.DataFrame(all_mdd_correlations).drop(index='MDD', errors='ignore')
mean_abs_corr = corr_df.abs().mean(axis=1)
top10_traits  = mean_abs_corr.nlargest(10).index.tolist()

col_order = [f"{r}_{c}" for r in REGIONS for c in CONDITIONS]
col_order = [c for c in col_order if c in corr_df.columns]
top10_df  = corr_df.loc[top10_traits, col_order]

print("Top 10 traits:")
for i, t in enumerate(top10_traits, 1):
    print(f"  {i:2d}. {t:20s}  mean|r| = {mean_abs_corr[t]:.3f}")

# ── 2. 图①热图：让色差更明显 ─────────────────────────────────────
# 关键：用实际数据范围而非固定 vmin/vmax
vmin = top10_df.values.min()
vmax = top10_df.values.max()
vcenter = (vmin + vmax) / 2
print(f"\n热图数值范围: {vmin:.3f} ~ {vmax:.3f}")

fig, ax = plt.subplots(figsize=(14, 5))
sns.heatmap(
    top10_df,
    cmap="RdYlBu_r",          # 红-黄-蓝，中间黄色过渡更易分辨
    center=vcenter,            # 以数据中值为中心
    vmin=vmin, vmax=vmax,      # 用真实范围撑开色阶
    linewidths=0.4, linecolor="white",
    cbar_kws={"label": "Pearson r with MDD", "shrink": 0.7},
    ax=ax
)

n_cond = len(CONDITIONS)
for i, region in enumerate(REGIONS):
    x = i * n_cond
    if i > 0:
        ax.axvline(x, color='black', linewidth=1.5)
    ax.text(x + n_cond / 2, -0.6, region,
            ha='center', va='bottom', fontsize=11,
            fontweight='bold', transform=ax.get_xaxis_transform())

ax.set_xticklabels([c.split("_")[1] for c in col_order],
                   rotation=45, ha='right', fontsize=8)
ax.set_yticklabels(ax.get_yticklabels(), fontsize=9)
ax.set_xlabel("")
ax.set_title("Top 10 MDD-correlated Traits — Pearson r across Regions & Conditions",
             fontsize=12, fontweight='bold', pad=18)
plt.tight_layout()
plt.savefig("top10_MDD_heatmap_overview.pdf", dpi=300, bbox_inches='tight')
plt.show()

# ── 3. 图②折线图：改为偏移 + 图例外置 ──────────────────────────
# 五条线贴在一起 → 用 z-score 标准化，看相对变化而非绝对值
region_colors = {
    "mPFC": "#E41A1C",
    "DRN":  "#377EB8",
    "Stri": "#4DAF4A",
    "HIP":  "#FF7F00",
    "VTA":  "#984EA3",
}

# 对每个 Trait × Region，做 Condition 方向的 z-score，突出条件间变化
def zscore_row(arr):
    s = np.std(arr)
    return (arr - np.mean(arr)) / (s + 1e-8)

fig, axes = plt.subplots(2, 5, figsize=(18, 7),
                         sharey=False,           # 各子图独立 y 轴，避免挤压
                         gridspec_kw={'hspace': 0.55, 'wspace': 0.35})
axes = axes.flatten()

for ax, trait in zip(axes[:10], top10_traits):
    for region in REGIONS:
        ys_raw = np.array([
            top10_df.loc[trait, f"{region}_{c}"]
            if f"{region}_{c}" in top10_df.columns else np.nan
            for c in CONDITIONS
        ])
        ys = zscore_row(ys_raw)   # 标准化到均值0，让各脑区线条散开
        ax.plot(range(len(CONDITIONS)), ys,
                marker='o', markersize=4, linewidth=1.5,
                color=region_colors[region], label=region)

    ax.axhline(0, linestyle='--', linewidth=0.6, color='gray')
    ax.set_xticks(range(len(CONDITIONS)))
    ax.set_xticklabels(CONDITIONS, rotation=40, ha='right', fontsize=7.5)
    ax.set_title(trait, fontsize=9, fontweight='bold')
    ax.set_ylabel("z-score", fontsize=7)
    ax.tick_params(axis='y', labelsize=7)

# 最后一格：只放图例，不显示轴
handles = [plt.Line2D([0], [0], color=c, marker='o',
                      linewidth=1.5, markersize=5, label=r)
           for r, c in region_colors.items()]

# 图例放在整张图右侧，完全不遮挡子图
fig.legend(handles=handles,
           title='Region', title_fontsize=9,
           fontsize=9, frameon=False,
           loc='center right',
           bbox_to_anchor=(1.0, 0.5))   # 贴右边缘

# 隐藏第11格（如果 top10 恰好10个）
for i in range(len(top10_traits), len(axes)):
    axes[i].set_visible(False)

fig.suptitle("Top 10 MDD-correlated Traits — z-scored Trajectory by Condition & Region",
             fontsize=12, fontweight='bold')
plt.savefig("top10_MDD_lineplot.pdf", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import os
import scanpy as sc
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ranksums

def load_ldsc(region, condition, trait):
    dfs = []
    for rep in ["rep1", "rep2"]:
        file_path = f"/biostack/home/suchenyi/gsmap/{region}_{condition}_{rep}/spatial_ldsc/{region}_{condition}_{rep}_{trait}.csv.gz"
        if not os.path.exists(file_path):
            continue

        df = pd.read_csv(file_path)
        df["spot"] = df["spot"].astype(str)
        df = df[["spot", "z"]].rename(columns={"z": f"z_{rep}"})
        dfs.append(df)

    if len(dfs) == 0:
        return None

    df_merge = dfs[0]
    for df in dfs[1:]:
        df_merge = df_merge.merge(df, on="spot", how="outer")

    z_cols = [c for c in df_merge.columns if "z_" in c]
    df_merge["z"] = df_merge[z_cols].mean(axis=1)

    return df_merge[["spot", "z"]]

def analyze_celltype_z(region, condition, trait="MDD"):

    print(f"Processing {region} - {condition}")

    h5ad_path = f"/biostack/home/suchenyi/gsmap/{region}_{condition}_rep1/{region}_{condition}_rep1_annotated.h5ad"
    if not os.path.exists(h5ad_path):
        return None

    adata = sc.read_h5ad(h5ad_path)
    adata.obs.index = adata.obs.index.astype(str)
    adata.obs["spot_clean"] = adata.obs.index.str.replace("-1", "", regex=False)

    df_ldsc = load_ldsc(region, condition, trait)
    if df_ldsc is None:
        return None

    df = adata.obs.merge(
        df_ldsc,
        left_on="spot_clean",
        right_on="spot",
        how="inner"
    )

    if df.empty:
        return None

    # ===== 提取细胞类型
    df["celltype"] = df["predicted_labels_plot"]

    return df[["celltype", "z"]]

REGION = "mPFC"
TRAIT = "MDD"
CONDITIONS = ["Control", "Sus", "Res","3D","7D"]

for condition in CONDITIONS:

    df = analyze_celltype_z(REGION, condition, TRAIT)
    if df is None:
        continue


    mask = df["celltype"].str.contains("Sst|Pvalb|Vip", case=False, na=False)
    df_sub = df.copy()
    #df_sub = df[mask].copy()
        #只保留 GABA interneuron

    if df_sub.empty:
        continue

    # ===== 画图
    plt.figure(figsize=(6,4))
    order = df.groupby("celltype")["z"].mean().sort_values().index


    sns.violinplot(
        data=df_sub,
        x="celltype",
        y="z",
        inner="box",
        cut=0,
        order=order
    )

    plt.title(f"{REGION} - {condition}: {TRAIT}")
    plt.xticks(rotation=90)
    plt.ylabel("z-score (spatial LDSC)")
    plt.tight_layout()

    plt.savefig(f"{REGION}_{condition}_violin.png", dpi=300)
    plt.show()

    # ===== 统计检验（SST vs others）
    if "Sst" in "".join(df_sub["celltype"].unique()):

        sst = df_sub[df_sub["celltype"].str.contains("Sst")]["z"]
        others = df_sub[~df_sub["celltype"].str.contains("Sst")]["z"]

        stat, p = ranksums(sst, others)

        print(f"{condition} | SST vs Others: p = {p:.3e}")
```

```python
import os
import scanpy as sc
import pandas as pd
import numpy as np

BASE_DIR = "/biostack/home/suchenyi/gsmap"

REGION = "mPFC"
TRAIT = "MDD"

# 定义新的时间顺序
PLOT_TIMES = ["Control", "3D", "7D", "Final"]

records = []

# 1. 处理共有时间点 (Control, 3D, 7D)
for time in ["Control", "3D", "7D"]:
    z_vals = get_celltype_z(REGION, time, cell_keyword="SST")
    if z_vals is not None:
        records.append({
            "group": "Shared", # 共有路径
            "time": time,
            "mean_z": np.mean(z_vals),
            "sem": np.std(z_vals) / np.sqrt(len(z_vals))
        })

# 2. 处理分化终点 (Sus, Res)
for group in ["Sus", "Res"]:
    z_vals = get_celltype_z(REGION, group, cell_keyword="SST")
    if z_vals is not None:
        records.append({
            "group": group,
            "time": "Final", # 统一标记为 Final 阶段
            "mean_z": np.mean(z_vals),
            "sem": np.std(z_vals) / np.sqrt(len(z_vals))
        })

df_plot = pd.DataFrame(records)

# 3. 绘图逻辑
plt.figure(figsize=(8, 5))

# 绘制共有路径 (Control -> 3D -> 7D)
shared = df_plot[df_plot["group"] == "Shared"].set_index("time").loc[["Control", "3D", "7D"]]
plt.errorbar(shared.index, shared["mean_z"], yerr=shared["sem"], fmt='o-', color='gray', label='Common Stress Path')

# 绘制从 7D 到 Sus 的连线
sus_final = df_plot[df_plot["group"] == "Sus"]
last_shared = shared.iloc[-1] # 7D 的数据
plt.errorbar(["7D", "Final"], [last_shared["mean_z"], sus_final["mean_z"].values[0]], 
             yerr=[last_shared["sem"], sus_final["sem"].values[0]], fmt='o--', color='red', label='Susceptible')

# 绘制从 7D 到 Res 的连线
res_final = df_plot[df_plot["group"] == "Res"]
plt.errorbar(["7D", "Final"], [last_shared["mean_z"], res_final["mean_z"].values[0]], 
             yerr=[last_shared["sem"], res_final["sem"].values[0]], fmt='o--', color='blue', label='Resilient')

plt.xticks(["Control", "3D", "7D", "Final"], ["Control", "3D", "7D", "Sus/Res"])
plt.ylabel("Mean z-score (SST, mPFC)")
plt.title("Temporal Trajectory of Genetic Susceptibility")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig("Y_trajectory_SST.png", dpi=300)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ── 1. Load all sheets ────────────────────────────────────────────
def load_data(filepath):
    xl = pd.ExcelFile(filepath)
    all_data = []

    for sheet in xl.sheet_names:
        parts = sheet.split("_", 1)  # split on first underscore only
        if len(parts) != 2:
            print(f"  ✗ Skipped sheet '{sheet}' (unexpected name format)")
            continue
        region, condition = parts[0], parts[1]

        df = pd.read_excel(filepath, sheet_name=sheet)
        df.rename(columns={df.columns[0]: 'Trait'}, inplace=True)
        df['Region'] = region
        df['Condition'] = condition
        all_data.append(df)
        print(f"  ✓ {sheet}: {len(df)} rows")

    return pd.concat(all_data, ignore_index=True)

df_all = load_data("Brain_Region_Trait_Clustering_Results.xlsx")
print(f"\nTotal rows: {len(df_all)}")
print(df_all.head())

# ── 2. Analysis A: Cross-region consistency heatmap (Sus group) ───
sus_df = df_all[df_all['Condition'] == 'Sus']
pivot_reg = sus_df.pivot(index='Trait', columns='Region', values='Is_MDD_Partner').astype(int)

plt.figure(figsize=(10, 15))
sns.heatmap(
    pivot_reg,
    cmap="Blues",
    cbar_kws={'label': 'Is MDD Partner (1=Yes, 0=No)'},
    linewidths=0.3,
    linecolor='lightgrey'
)
plt.title("Cross-Region Consistency in Susceptible Group", fontsize=14)
plt.tight_layout()
plt.savefig('region_consistency.png', dpi=150)
plt.show()
print("Saved: region_consistency.png")

# ── 3. Analysis B: Condition-transition heatmap (mPFC region) ─────
mpfc_df = df_all[df_all['Region'] == 'mPFC'].copy()

cond_order = ['Control', '3D', '7D', 'Sus', 'Res']
mpfc_df['Condition'] = pd.Categorical(mpfc_df['Condition'], categories=cond_order, ordered=True)
pivot_cond = mpfc_df.pivot(index='Trait', columns='Condition', values='Is_MDD_Partner').astype(int)

plt.figure(figsize=(8, 15))
sns.heatmap(
    pivot_cond,
    cmap="Reds",
    cbar_kws={'label': 'Is MDD Partner (1=Yes, 0=No)'},
    linewidths=0.3,
    linecolor='lightgrey'
)
plt.title("Trait-MDD Clustering Transition in mPFC", fontsize=14)
plt.tight_layout()
plt.savefig('condition_transition.png', dpi=150)
plt.show()
print("Saved: condition_transition.png")
```

```python
import pandas as pd
import plotly.graph_objects as go

# ── 1. 准备 mPFC 数据 ─────────────────────────────────────────────
df_mpfc = df_all[df_all['Region'] == 'mPFC'].copy()

cond_order = ['Control', '3D', '7D', 'Sus']
df_mpfc['Condition'] = pd.Categorical(df_mpfc['Condition'], categories=cond_order, ordered=True)

# 宽表：每行一个 Trait，每列一个 Condition，值为 Is_MDD_Partner (bool)
wide = df_mpfc.pivot(index='Trait', columns='Condition', values='Is_MDD_Partner')
wide = wide[cond_order].dropna()  # 保留在所有条件都有数据的 Trait

# ── 2. 构建桑基图节点和流 ─────────────────────────────────────────
# 节点命名规则: "Condition_状态", 例如 "Control_MDD", "3D_nonMDD"
def node_label(cond, is_mdd):
    return f"{cond}_{'MDD' if is_mdd else 'non-MDD'}"

# 所有节点（每个 Condition × 2种状态 = 10个节点）
nodes = []
for cond in cond_order:
    nodes.append(node_label(cond, True))
    nodes.append(node_label(cond, False))
node_idx = {n: i for i, n in enumerate(nodes)}

# 统计相邻条件之间的流量，并记录流经的 Traits
source_list, target_list, value_list, trait_list = [], [], [], []

for i in range(len(cond_order) - 1):
    c1, c2 = cond_order[i], cond_order[i + 1]
    for mdd1 in [True, False]:
        for mdd2 in [True, False]:
            traits = wide[
                (wide[c1] == mdd1) & (wide[c2] == mdd2)
            ].index.tolist()
            if traits:
                source_list.append(node_idx[node_label(c1, mdd1)])
                target_list.append(node_idx[node_label(c2, mdd2)])
                value_list.append(len(traits))
                trait_list.append("<br>".join(traits[:30]))  # 最多显示30个

# ── 3. 节点颜色 ───────────────────────────────────────────────────
node_colors = []
for n in nodes:
    if "MDD" in n and "non" not in n:
        node_colors.append("rgba(214, 39, 40, 0.8)")    # 红：MDD partner
    else:
        node_colors.append("rgba(158, 202, 225, 0.8)")  # 蓝：non-MDD

# 流的颜色：由起始节点状态决定
link_colors = []
for src in source_list:
    if "MDD" in nodes[src] and "non" not in nodes[src]:
        link_colors.append("rgba(214, 39, 40, 0.25)")
    else:
        link_colors.append("rgba(158, 202, 225, 0.25)")

# ── 4. 绘图 ───────────────────────────────────────────────────────
fig = go.Figure(go.Sankey(
    arrangement="snap",
    node=dict(
        pad=20,
        thickness=25,
        line=dict(color="white", width=0.5),
        label=nodes,
        color=node_colors,
        hovertemplate="%{label}<br>Trait 数量: %{value}<extra></extra>"
    ),
    link=dict(
        source=source_list,
        target=target_list,
        value=value_list,
        color=link_colors,
        customdata=trait_list,
        hovertemplate=(
            "%{source.label} → %{target.label}<br>"
            "数量: %{value}<br><br>"
            "Traits:<br>%{customdata}<extra></extra>"
        )
    )
))

fig.update_layout(
    title=dict(
        text="mPFC | Trait-MDD 聚类状态流转 (Control → 3D → 7D → Sus)",
        font_size=15
    ),
    font_size=11,
    width=1000,
    height=600,
    paper_bgcolor="white"
)

fig.write_html("sankey_mPFC_sus.html")  # 保存为可交互 HTML
fig.show()
print("Saved: sankey_mPFC.html")
```
